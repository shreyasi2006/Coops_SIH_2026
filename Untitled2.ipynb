```python
def haversine_distance(lat1, lon1, lat2, lon2):

    R = 6371  # Earth radius in kilometres

    lat1 = np.radians(lat1)
    lat2 = np.radians(lat2)

    dlat = lat2 - lat1
    dlon = np.radians(lon2) - np.radians(lon1)

    a = (
        np.sin(dlat / 2) ** 2
        +
        np.cos(lat1)
        * np.cos(lat2)
        * np.sin(dlon / 2) ** 2
    )

    c = 2 * np.arcsin(np.sqrt(a))

    return R * c
```

```python
MAX_DISTANCE = 50

def distance_score(distance):

    score = 1 - (distance / MAX_DISTANCE)

    return max(0, min(1, score))
```

```python
def availability_score(availability):

    if str(availability).lower() == "available":
        return 1.0

    return 0.0
```

```python
def rating_score(rating):

    return min(float(rating) / 5.0, 1.0)
```

```python
def experience_score(experience):

    return min(float(experience) / 10.0, 1.0)
```

```python
def certification_score(certification):

    if str(certification).lower() == "certified":
        return 1.0

    return 0.0
```

```python
def category_match_score(worker_category, request_category):

    if str(worker_category).lower() == str(request_category).lower():
        return 1.0

    return 0.0
```

```python
def calculate_matching_score(worker, request):

    distance = haversine_distance(
        request["customer_latitude"],
        request["customer_longitude"],
        worker["latitude"],
        worker["longitude"]
    )

    skill = category_match_score(
        worker["service_category"],
        request["service_category"]
    )

    distance_s = distance_score(distance)
    availability_s = availability_score(worker["availability"])
    rating_s = rating_score(worker["rating"])
    experience_s = experience_score(worker["experience_years"])
    certification_s = certification_score(worker["certification"])

    final_score = (
        skill * 0.30
        + distance_s * 0.20
        + availability_s * 0.20
        + rating_s * 0.15
        + experience_s * 0.10
        + certification_s * 0.05
    )

    return {
        "skill_score": skill,
        "distance_km": distance,
        "distance_score": distance_s,
        "availability_score": availability_s,
        "rating_score": rating_s,
        "experience_score": experience_s,
        "certification_score": certification_s,
        "final_score": final_score
    }
```

```python
from scipy.spatial.distance import euclidean

def calculate_skill_match(worker, service_category):

    if service_category not in service_profiles:
        return 0.0

    target = np.array([
        service_profiles[service_category][feature]
        for feature in onet_features
    ])

    worker_profile = np.array([
        worker[feature]
        for feature in onet_features
    ])

    distance = euclidean(
        target,
        worker_profile
    )

    # Maximum possible distance in 5-dimensional
    # normalized space
    max_distance = np.sqrt(5)

    similarity = 1 - (
        distance / max_distance
    )

    return max(0, min(1, similarity))
```

```python
import pandas as pd
import numpy as np

workers = pd.read_csv(
    "/content/coopwork_workers_with_skills.csv"
)

requests = pd.read_csv(
    "/content/customer_requests.csv"
)

print("Workers:", workers.shape)
print("Requests:", requests.shape)

display(workers.head())
display(requests.head())
```

```python
required_worker_columns = [
    "worker_id",
    "service_category",
    "experience_years",
    "rating",
    "availability",
    "latitude",
    "longitude",
    "certification",
    "verification_status",
    "skill_match_score"
]

required_request_columns = [
    "request_id",
    "service_category",
    "customer_latitude",
    "customer_longitude",
    "preferred_time",
    "urgency"
]

print("Missing worker columns:")
print([
    c for c in required_worker_columns
    if c not in workers.columns
])

print("\nMissing request columns:")
print([
    c for c in required_request_columns
    if c not in requests.columns
])
```

```python
pairs = []

for _, request in requests.iterrows():

    candidates = workers[
        workers["service_category"].str.lower()
        == request["service_category"].lower()
    ]

    for _, worker in candidates.iterrows():

        pairs.append({
            "request_id": request["request_id"],
            "worker_id": worker["worker_id"],
            "service_category": request["service_category"],
            "skill_match_score": worker["skill_match_score"],
            "experience_years": worker["experience_years"],
            "rating": worker["rating"],
            "availability": worker["availability"],
            "certification": worker["certification"],
            "worker_latitude": worker["latitude"],
            "worker_longitude": worker["longitude"],
            "customer_latitude": request["customer_latitude"],
            "customer_longitude": request["customer_longitude"],
            "preferred_time": request["preferred_time"],
            "urgency": request["urgency"]
        })

matching_df = pd.DataFrame(pairs)

print("Matching pairs:", matching_df.shape)

display(matching_df.head(10))
```

```python
def haversine_distance(lat1, lon1, lat2, lon2):

    R = 6371

    lat1 = np.radians(lat1)
    lat2 = np.radians(lat2)

    dlat = lat2 - lat1
    dlon = np.radians(lon2) - np.radians(lon1)

    a = (
        np.sin(dlat / 2) ** 2
        + np.cos(lat1)
        * np.cos(lat2)
        * np.sin(dlon / 2) ** 2
    )

    return 2 * R * np.arcsin(np.sqrt(a))


matching_df["distance_km"] = matching_df.apply(
    lambda row: haversine_distance(
        row["customer_latitude"],
        row["customer_longitude"],
        row["worker_latitude"],
        row["worker_longitude"]
    ),
    axis=1
)

matching_df["distance_score"] = (
    1 - matching_df["distance_km"] / 50
).clip(0, 1)

display(
    matching_df[
        [
            "request_id",
            "worker_id",
            "distance_km",
            "distance_score"
        ]
    ].head(10)
)
```

```python
matching_df["availability_score"] = (
    matching_df["availability"]
    .str.lower()
    .eq("available")
    .astype(float)
)

matching_df["rating_score"] = (
    matching_df["rating"] / 5
).clip(0, 1)

matching_df["experience_score"] = (
    matching_df["experience_years"] / 10
).clip(0, 1)

matching_df["certification_score"] = (
    matching_df["certification"]
    .str.lower()
    .eq("certified")
    .astype(float)
)
```

```python
matching_df["final_match_score"] = (
    matching_df["skill_match_score"] * 0.30
    + matching_df["distance_score"] * 0.20
    + matching_df["availability_score"] * 0.20
    + matching_df["rating_score"] * 0.15
    + matching_df["experience_score"] * 0.10
    + matching_df["certification_score"] * 0.05
)

matching_df["final_match_percent"] = (
    matching_df["final_match_score"] * 100
).round(2)
```

```python
top_matches = matching_df.sort_values(
    "final_match_score",
    ascending=False
)

display(
    top_matches[
        [
            "request_id",
            "worker_id",
            "service_category",
            "skill_match_score",
            "distance_km",
            "availability_score",
            "rating_score",
            "experience_score",
            "certification_score",
            "final_match_percent"
        ]
    ].head(20)
)
```

```python
def assign_match_class(score):

    if score >= 80:
        return "Excellent"

    elif score >= 65:
        return "Good"

    elif score >= 50:
        return "Moderate"

    else:
        return "Poor"


matching_df["match_class"] = (
    matching_df["final_match_percent"]
    .apply(assign_match_class)
)

display(
    matching_df[
        [
            "request_id",
            "worker_id",
            "final_match_percent",
            "match_class"
        ]
    ].head(20)
)
```

```python
# ============================================================
# STEP 6: RANDOM FOREST TRAINING + EVALUATION
# CORRECTED VERSION
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import joblib

from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay
)

# ============================================================
# 1. LOAD DATASET
# ============================================================

file_path = "/content/matching_training_dataset_labelled.csv"

df = pd.read_csv(file_path)

print("Dataset Shape:", df.shape)
print("\nColumns available:")
print(df.columns.tolist())


# ============================================================
# 2. DEFINE FEATURES
# ============================================================

features = [
    "skill_match_score",
    "distance_score",
    "availability_score",
    "rating_score",
    "experience_score",
    "certification_score"
]


# ============================================================
# 3. CHECK FEATURES
# ============================================================

missing_features = [
    col for col in features
    if col not in df.columns
]

if missing_features:
    print("\nMissing feature columns:")
    print(missing_features)
    raise ValueError(
        "Some required feature columns are missing."
    )

print("\nAll required feature columns are present.")


# ============================================================
# 4. CREATE TARGET COLUMN
# ============================================================

# If match_class_num already exists
if "match_class_num" in df.columns:

    print("\nUsing existing match_class_num column.")

# If match_class exists, convert it to numbers
elif "match_class" in df.columns:

    print("\nCreating match_class_num from match_class...")

    class_mapping = {
        "Poor": 0,
        "Moderate": 1,
        "Good": 2,
        "Excellent": 3
    }

    # Remove unwanted spaces/capitalization
    df["match_class"] = (
        df["match_class"]
        .astype(str)
        .str.strip()
        .str.title()
    )

    df["match_class_num"] = df["match_class"].map(
        class_mapping
    )

# Otherwise create class from final_match_percent
elif "final_match_percent" in df.columns:

    print(
        "\nmatch_class not found."
        " Creating classes from final_match_percent..."
    )

    def create_class(score):

        if score >= 80:
            return 3
        elif score >= 65:
            return 2
        elif score >= 50:
            return 1
        else:
            return 0

    df["match_class_num"] = (
        df["final_match_percent"]
        .apply(create_class)
    )

else:

    raise ValueError(
        "\nCannot create target column.\n"
        "Dataset must contain either:\n"
        "1. match_class_num\n"
        "2. match_class\n"
        "3. final_match_percent"
    )


# ============================================================
# 5. REMOVE INVALID TARGET ROWS
# ============================================================

df = df.dropna(
    subset=features + ["match_class_num"]
)

# Convert target to integer
df["match_class_num"] = (
    df["match_class_num"]
    .astype(int)
)

print("\nFinal dataset shape:", df.shape)


# ============================================================
# 6. DISPLAY CLASS DISTRIBUTION
# ============================================================

class_names = {
    0: "Poor",
    1: "Moderate",
    2: "Good",
    3: "Excellent"
}

print("\nMatch Class Distribution:")

for class_id, class_name in class_names.items():

    count = (
        df["match_class_num"] == class_id
    ).sum()

    print(
        f"{class_id} - {class_name}: {count}"
    )


# ============================================================
# 7. CREATE X AND Y
# ============================================================

X = df[features]
y = df["match_class_num"]


# ============================================================
# 8. TRAIN-TEST SPLIT
# ============================================================

X_train, X_test, y_train, y_test = train_test_split(

    X,
    y,

    test_size=0.20,

    random_state=42,

    stratify=y
)

print("\nTraining samples:", len(X_train))
print("Testing samples :", len(X_test))


# ============================================================
# 9. CREATE RANDOM FOREST
# ============================================================

rf_model = RandomForestClassifier(

    n_estimators=200,

    max_depth=12,

    min_samples_split=5,

    min_samples_leaf=2,

    random_state=42,

    class_weight="balanced"
)


# ============================================================
# 10. TRAIN RANDOM FOREST
# ============================================================

rf_model.fit(
    X_train,
    y_train
)

print("\n======================================")
print("RANDOM FOREST TRAINING COMPLETED")
print("======================================")


# ============================================================
# 11. PREDICTION
# ============================================================

y_pred = rf_model.predict(
    X_test
)


# ============================================================
# 12. EVALUATION
# ============================================================

accuracy = accuracy_score(
    y_test,
    y_pred
)

precision = precision_score(
    y_test,
    y_pred,
    average="weighted",
    zero_division=0
)

recall = recall_score(
    y_test,
    y_pred,
    average="weighted",
    zero_division=0
)

f1 = f1_score(
    y_test,
    y_pred,
    average="weighted",
    zero_division=0
)


print("\n======================================")
print("       RANDOM FOREST RESULTS")
print("======================================")

print(f"Accuracy  : {accuracy * 100:.2f}%")
print(f"Precision : {precision * 100:.2f}%")
print(f"Recall    : {recall * 100:.2f}%")
print(f"F1-Score  : {f1 * 100:.2f}%")


# ============================================================
# 13. CLASSIFICATION REPORT
# ============================================================

print("\n======================================")
print("CLASSIFICATION REPORT")
print("======================================")

print(
    classification_report(
        y_test,
        y_pred,
        labels=[0, 1, 2, 3],
        target_names=[
            "Poor",
            "Moderate",
            "Good",
            "Excellent"
        ],
        zero_division=0
    )
)


# ============================================================
# 14. CONFUSION MATRIX
# ============================================================

cm = confusion_matrix(
    y_test,
    y_pred,
    labels=[0, 1, 2, 3]
)

print("\nConfusion Matrix:")
print(cm)

fig, ax = plt.subplots(
    figsize=(7, 6)
)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=[
        "Poor",
        "Moderate",
        "Good",
        "Excellent"
    ]
)

disp.plot(
    ax=ax,
    cmap="Blues",
    values_format="d"
)

plt.title(
    "Random Forest - Confusion Matrix"
)

plt.xlabel(
    "Predicted Class"
)

plt.ylabel(
    "Actual Class"
)

plt.show()


# ============================================================
# 15. FEATURE IMPORTANCE
# ============================================================

importance = pd.DataFrame({

    "Feature": features,

    "Importance":
        rf_model.feature_importances_

})

importance = importance.sort_values(
    by="Importance",
    ascending=False
)

print("\n======================================")
print("FEATURE IMPORTANCE")
print("======================================")

display(importance)


# ============================================================
# 16. FEATURE IMPORTANCE GRAPH
# ============================================================

plt.figure(
    figsize=(9, 5)
)

plt.bar(
    importance["Feature"],
    importance["Importance"]
)

plt.title(
    "Random Forest Feature Importance"
)

plt.xlabel(
    "Features"
)

plt.ylabel(
    "Importance"
)

plt.xticks(
    rotation=45,
    ha="right"
)

plt.tight_layout()

plt.show()


# ============================================================
# 17. SAVE RANDOM FOREST MODEL
# ============================================================

model_path = (
    "/content/random_forest_match_model.pkl"
)

joblib.dump(
    rf_model,
    model_path
)

print(
    "\nRandom Forest model saved successfully:"
)

print(model_path)


# ============================================================
# 18. SAVE TEST RESULTS
# ============================================================

results = X_test.copy()

results["actual_class"] = y_test.values

results["predicted_class"] = y_pred

results["actual_label"] = (
    results["actual_class"]
    .map(class_names)
)

results["predicted_label"] = (
    results["predicted_class"]
    .map(class_names)
)

results.to_csv(
    "/content/random_forest_test_results.csv",
    index=False
)

print(
    "\nTest results saved successfully:"
)

print(
    "/content/random_forest_test_results.csv"
)


# ============================================================
# 19. FINAL STATUS
# ============================================================

print("\n======================================")
print("STEP 6 COMPLETED SUCCESSFULLY")
print("======================================")
```

```python
# ============================================================
# STEP 7: AI TOP-5 WORKER RECOMMENDATION ENGINE
# FIXED VERSION - NO 'skills' COLUMN REQUIRED
# ============================================================

import pandas as pd
import numpy as np
import joblib

# ============================================================
# 1. LOAD DATA
# ============================================================

workers = pd.read_csv(
    "/content/coopwork_workers.csv"
)

requests = pd.read_csv(
    "/content/customer_requests.csv"
)

model = joblib.load(
    "/content/random_forest_match_model.pkl"
)

print("Workers dataset shape :", workers.shape)
print("Requests dataset shape:", requests.shape)

print("\nWorker columns:")
print(workers.columns.tolist())

print("\nRequest columns:")
print(requests.columns.tolist())


# ============================================================
# 2. REQUIRED MODEL FEATURES
# ============================================================

features = [
    "skill_match_score",
    "distance_score",
    "availability_score",
    "rating_score",
    "experience_score",
    "certification_score"
]


# ============================================================
# 3. CLASS LABELS
# ============================================================

class_names = {
    0: "Poor",
    1: "Moderate",
    2: "Good",
    3: "Excellent"
}


# ============================================================
# 4. SERVICE-SPECIFIC SKILL PROFILES
# ============================================================

service_skills = {

    "Electrical": [
        "wiring",
        "electrical repair",
        "installation",
        "circuit repair"
    ],

    "Plumbing": [
        "pipe repair",
        "water supply",
        "leak repair",
        "installation"
    ],

    "Carpentry": [
        "woodwork",
        "furniture repair",
        "furniture installation",
        "measurement"
    ],

    "Painting": [
        "wall painting",
        "surface preparation",
        "color application",
        "finishing"
    ],

    "Cleaning": [
        "cleaning",
        "sanitation",
        "housekeeping",
        "waste handling"
    ],

    "AC Repair": [
        "AC servicing",
        "cooling system",
        "maintenance",
        "troubleshooting"
    ],

    "Appliance Repair": [
        "appliance repair",
        "troubleshooting",
        "maintenance",
        "installation"
    ],

    "Gardening": [
        "plant care",
        "gardening",
        "landscaping",
        "maintenance"
    ],

    "Construction": [
        "construction",
        "building work",
        "measurement",
        "site work"
    ],

    "Masonry": [
        "brickwork",
        "cement work",
        "construction",
        "wall repair"
    ],

    "Welding": [
        "welding",
        "metal fabrication",
        "metal repair",
        "cutting"
    ],

    "Vehicle Repair": [
        "vehicle maintenance",
        "engine repair",
        "mechanical repair",
        "troubleshooting"
    ]
}


# ============================================================
# 5. HELPER FUNCTION
# ============================================================

def normalize_text(value):

    if pd.isna(value):
        return ""

    return str(value).strip().lower()


# ============================================================
# 6. TOP-5 RECOMMENDATION FUNCTION
# ============================================================

def recommend_workers(request_id, top_n=5):

    # --------------------------------------------------------
    # Find customer request
    # --------------------------------------------------------

    request_rows = requests[
        requests["request_id"] == request_id
    ]

    if request_rows.empty:

        print("Request ID not found.")

        return None

    request = request_rows.iloc[0]

    category = str(
        request["service_category"]
    ).strip()


    # --------------------------------------------------------
    # Display request
    # --------------------------------------------------------

    print("\n==========================================")
    print("CUSTOMER REQUEST")
    print("==========================================")

    print(
        "Request ID :",
        request["request_id"]
    )

    print(
        "Service    :",
        category
    )

    print(
        "Required Skills:",
        request["required_skills"]
    )

    print(
        "Urgency    :",
        request["urgency"]
    )


    # --------------------------------------------------------
    # Filter workers by service category
    # --------------------------------------------------------

    candidates = workers[
        workers["service_category"]
        .astype(str)
        .str.strip()
        .str.lower()
        ==
        category.lower()
    ].copy()


    if candidates.empty:

        print(
            "\nNo workers found for this service category."
        )

        return None


    print(
        "\nCandidates found:",
        len(candidates)
    )


    # ========================================================
    # 7. DISTANCE CALCULATION
    # ========================================================

    R = 6371.0

    customer_lat = float(
        request["customer_latitude"]
    )

    customer_lon = float(
        request["customer_longitude"]
    )

    lat1 = np.radians(customer_lat)
    lon1 = np.radians(customer_lon)

    lat2 = np.radians(
        pd.to_numeric(
            candidates["latitude"],
            errors="coerce"
        )
    )

    lon2 = np.radians(
        pd.to_numeric(
            candidates["longitude"],
            errors="coerce"
        )
    )

    dlat = lat2 - lat1
    dlon = lon2 - lon1

    a = (
        np.sin(dlat / 2) ** 2
        +
        np.cos(lat1)
        * np.cos(lat2)
        * np.sin(dlon / 2) ** 2
    )

    a = np.clip(a, 0, 1)

    c = 2 * np.arctan2(
        np.sqrt(a),
        np.sqrt(1 - a)
    )

    candidates["distance_km"] = R * c


    # ========================================================
    # 8. DISTANCE SCORE
    # ========================================================

    max_distance = candidates[
        "distance_km"
    ].max()

    if pd.isna(max_distance) or max_distance == 0:

        candidates["distance_score"] = 1.0

    else:

        candidates["distance_score"] = (
            1
            -
            candidates["distance_km"]
            / max_distance
        )

        candidates["distance_score"] = (
            candidates["distance_score"]
            .clip(0, 1)
        )


    # ========================================================
    # 9. AVAILABILITY SCORE
    # ========================================================

    availability_text = (
        candidates["availability"]
        .astype(str)
        .str.strip()
        .str.lower()
    )

    candidates["availability_score"] = (
        availability_text
        .map({
            "available": 1.0,
            "unavailable": 0.0,
            "true": 1.0,
            "false": 0.0,
            "yes": 1.0,
            "no": 0.0
        })
        .fillna(0.0)
    )


    # ========================================================
    # 10. RATING SCORE
    # ========================================================

    candidates["rating_score"] = (
        pd.to_numeric(
            candidates["rating"],
            errors="coerce"
        )
        / 5.0
    ).clip(0, 1).fillna(0)


    # ========================================================
    # 11. EXPERIENCE SCORE
    # ========================================================

    candidates["experience_years"] = pd.to_numeric(
        candidates["experience_years"],
        errors="coerce"
    ).fillna(0)

    max_exp = candidates[
        "experience_years"
    ].max()

    if max_exp > 0:

        candidates["experience_score"] = (
            candidates["experience_years"]
            / max_exp
        ).clip(0, 1)

    else:

        candidates["experience_score"] = 0.0


    # ========================================================
    # 12. CERTIFICATION SCORE
    # ========================================================

    certification_text = (
        candidates["certification"]
        .astype(str)
        .str.strip()
        .str.lower()
    )

    candidates["certification_score"] = (
        certification_text
        .map({
            "certified": 1.0,
            "not certified": 0.0,
            "yes": 1.0,
            "no": 0.0,
            "true": 1.0,
            "false": 0.0
        })
        .fillna(0.0)
    )


    # ========================================================
    # 13. SKILL MATCHING
    # ========================================================

    # Your worker dataset does not contain 'skills'.
    # Therefore we use service-category based skill matching.

    required_skills = str(
        request["required_skills"]
    ).lower()

    required_skills = [
        x.strip()
        for x in required_skills.split(",")
        if x.strip()
    ]


    def calculate_skill_match(worker_category):

        worker_category = str(
            worker_category
        ).strip()

        worker_profile = service_skills.get(
            worker_category,
            []
        )

        worker_profile = [
            x.lower()
            for x in worker_profile
        ]

        if len(required_skills) == 0:

            return 0.0

        matched = 0

        for required in required_skills:

            for worker_skill in worker_profile:

                if (
                    required in worker_skill
                    or
                    worker_skill in required
                ):

                    matched += 1

                    break

        return matched / len(required_skills)


    candidates["skill_match_score"] = (
        candidates["service_category"]
        .apply(calculate_skill_match)
    )


    # ========================================================
    # 14. PREPARE ML FEATURES
    # ========================================================

    X_new = candidates[
        features
    ].copy()

    X_new = X_new.replace(
        [np.inf, -np.inf],
        np.nan
    )

    X_new = X_new.fillna(0)


    # ========================================================
    # 15. RANDOM FOREST PREDICTION
    # ========================================================

    candidates["predicted_class"] = (
        model.predict(X_new)
    )


    candidates["match_class"] = (
        candidates["predicted_class"]
        .map(class_names)
    )


    # ========================================================
    # 16. MODEL CONFIDENCE
    # ========================================================

    if hasattr(model, "predict_proba"):

        probabilities = model.predict_proba(
            X_new
        )

        candidates["confidence"] = (
            np.max(
                probabilities,
                axis=1
            ) * 100
        )

    else:

        candidates["confidence"] = 0


    # ========================================================
    # 17. FINAL MATCH SCORE
    # ========================================================

    candidates["final_match_score"] = (

        0.30
        * candidates["skill_match_score"]

        +

        0.20
        * candidates["distance_score"]

        +

        0.20
        * candidates["availability_score"]

        +

        0.15
        * candidates["rating_score"]

        +

        0.10
        * candidates["experience_score"]

        +

        0.05
        * candidates["certification_score"]

    )


    candidates["match_percent"] = (
        candidates["final_match_score"]
        * 100
    )


    # ========================================================
    # 18. SORT TOP WORKERS
    # ========================================================

    top_workers = candidates.sort_values(
        by="final_match_score",
        ascending=False
    ).head(top_n)


    # ========================================================
    # 19. PREPARE OUTPUT
    # ========================================================

    output_columns = [
        "worker_id",
        "service_category",
        "experience_years",
        "rating",
        "availability",
        "distance_km",
        "certification",
        "verification_status",
        "skill_match_score",
        "match_percent",
        "match_class",
        "confidence"
    ]

    output = top_workers[
        output_columns
    ].copy()


    # Round numerical values

    output["distance_km"] = (
        output["distance_km"]
        .round(2)
    )

    output["skill_match_score"] = (
        output["skill_match_score"]
        .round(3)
    )

    output["match_percent"] = (
        output["match_percent"]
        .round(2)
    )

    output["confidence"] = (
        output["confidence"]
        .round(2)
    )


    # ========================================================
    # 20. DISPLAY TOP-5
    # ========================================================

    print("\n==========================================")
    print(
        f"TOP {top_n} AI RECOMMENDED WORKERS"
    )
    print("==========================================")

    display(output)


    return output


# ============================================================
# 21. TEST THE AI ENGINE
# ============================================================

sample_request_id = requests.iloc[0]["request_id"]

top5 = recommend_workers(
    sample_request_id,
    top_n=5
)
```

```python
# ============================================================
# STEP 8: EXPLAINABLE AI (XAI)
# Generate reasons for every recommended worker
# ============================================================

import pandas as pd
import numpy as np

# ============================================================
# 1. CHECK THAT TOP-5 RESULTS EXIST
# ============================================================

if "top5" not in globals() or top5 is None:

    raise ValueError(
        "Top-5 recommendation results not found.\n"
        "Please run Step 7 first."
    )

print("Step 7 results found successfully.")


# ============================================================
# 2. CREATE EXPLANATION FUNCTION
# ============================================================

def generate_explanation(row):

    reasons = []


    # --------------------------------------------------------
    # Skill Match
    # --------------------------------------------------------

    skill_score = float(
        row["skill_match_score"]
    )

    if skill_score >= 0.80:

        reasons.append(
            f"Strong skill match ({skill_score * 100:.0f}%)"
        )

    elif skill_score >= 0.50:

        reasons.append(
            f"Good skill match ({skill_score * 100:.0f}%)"
        )

    else:

        reasons.append(
            f"Partial skill match ({skill_score * 100:.0f}%)"
        )


    # --------------------------------------------------------
    # Distance
    # --------------------------------------------------------

    distance = float(
        row["distance_km"]
    )

    if distance <= 2:

        reasons.append(
            f"Very close ({distance:.2f} km)"
        )

    elif distance <= 5:

        reasons.append(
            f"Nearby ({distance:.2f} km)"
        )

    else:

        reasons.append(
            f"Distance: {distance:.2f} km"
        )


    # --------------------------------------------------------
    # Availability
    # --------------------------------------------------------

    availability = str(
        row["availability"]
    ).strip().lower()

    if availability == "available":

        reasons.append(
            "Currently available"
        )

    else:

        reasons.append(
            "Availability is limited"
        )


    # --------------------------------------------------------
    # Rating
    # --------------------------------------------------------

    rating = float(
        row["rating"]
    )

    if rating >= 4.5:

        reasons.append(
            f"Excellent rating ({rating:.1f}/5)"
        )

    elif rating >= 4.0:

        reasons.append(
            f"Good rating ({rating:.1f}/5)"
        )

    else:

        reasons.append(
            f"Rating: {rating:.1f}/5"
        )


    # --------------------------------------------------------
    # Experience
    # --------------------------------------------------------

    experience = float(
        row["experience_years"]
    )

    if experience >= 5:

        reasons.append(
            f"Experienced ({experience:.0f} years)"
        )

    elif experience >= 2:

        reasons.append(
            f"{experience:.0f} years experience"
        )

    else:

        reasons.append(
            "Entry-level experience"
        )


    # --------------------------------------------------------
    # Certification
    # --------------------------------------------------------

    certification = str(
        row["certification"]
    ).strip().lower()

    verification = str(
        row["verification_status"]
    ).strip().lower()


    if certification == "certified":

        reasons.append(
            "Certified"
        )

    if verification == "verified":

        reasons.append(
            "Cooperative verified"
        )


    return " • ".join(reasons)


# ============================================================
# 3. GENERATE EXPLANATIONS
# ============================================================

top5_explained = top5.copy()

top5_explained["explanation"] = (
    top5_explained.apply(
        generate_explanation,
        axis=1
    )
)


# ============================================================
# 4. CREATE CUSTOMER-FRIENDLY OUTPUT
# ============================================================

display_columns = [
    "worker_id",
    "service_category",
    "match_percent",
    "match_class",
    "distance_km",
    "rating",
    "experience_years",
    "availability",
    "certification",
    "verification_status",
    "explanation"
]

final_recommendations = top5_explained[
    display_columns
].copy()


# ============================================================
# 5. DISPLAY RESULTS
# ============================================================

print("\n==========================================")
print("      EXPLAINABLE AI RECOMMENDATIONS")
print("==========================================")

display(
    final_recommendations
)


# ============================================================
# 6. SAVE RESULTS
# ============================================================

output_path = (
    "/content/top5_explainable_recommendations.csv"
)

final_recommendations.to_csv(
    output_path,
    index=False
)

print(
    "\nExplainable recommendations saved to:"
)

print(output_path)


# ============================================================
# 7. DISPLAY INDIVIDUAL RECOMMENDATIONS
# ============================================================

print("\n==========================================")
print("       AI RECOMMENDATION SUMMARY")
print("==========================================")

for index, row in final_recommendations.iterrows():

    print(
        f"\n#{list(final_recommendations.index).index(index) + 1}"
    )

    print(
        f"Worker ID       : {row['worker_id']}"
    )

    print(
        f"Service         : {row['service_category']}"
    )

    print(
        f"Match Score     : {row['match_percent']:.2f}%"
    )

    print(
        f"Match Class     : {row['match_class']}"
    )

    print(
        f"Why Recommended : {row['explanation']}"
    )
```

```python
# ============================================================
# STEP 9: TEST AI ON MULTIPLE CUSTOMER REQUESTS
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# ============================================================
# 1. CHECK STEP 7 DATA
# ============================================================

if "workers" not in globals():
    workers = pd.read_csv(
        "/content/coopwork_workers.csv"
    )

if "requests" not in globals():
    requests = pd.read_csv(
        "/content/customer_requests.csv"
    )

if "model" not in globals():
    import joblib

    model = joblib.load(
        "/content/random_forest_match_model.pkl"
    )


# ============================================================
# 2. SERVICE SKILL PROFILES
# ============================================================

service_skills = {

    "Electrical": [
        "wiring",
        "electrical repair",
        "installation",
        "circuit repair"
    ],

    "Plumbing": [
        "pipe repair",
        "water supply",
        "leak repair",
        "installation"
    ],

    "Carpentry": [
        "woodwork",
        "furniture repair",
        "furniture installation",
        "measurement"
    ],

    "Painting": [
        "wall painting",
        "surface preparation",
        "color application",
        "finishing"
    ],

    "Cleaning": [
        "cleaning",
        "sanitation",
        "housekeeping",
        "waste handling"
    ],

    "AC Repair": [
        "AC servicing",
        "cooling system",
        "maintenance",
        "troubleshooting"
    ],

    "Appliance Repair": [
        "appliance repair",
        "troubleshooting",
        "maintenance",
        "installation"
    ],

    "Gardening": [
        "plant care",
        "gardening",
        "landscaping",
        "maintenance"
    ],

    "Construction": [
        "construction",
        "building work",
        "measurement",
        "site work"
    ],

    "Masonry": [
        "brickwork",
        "cement work",
        "construction",
        "wall repair"
    ],

    "Welding": [
        "welding",
        "metal fabrication",
        "metal repair",
        "cutting"
    ],

    "Vehicle Repair": [
        "vehicle maintenance",
        "engine repair",
        "mechanical repair",
        "troubleshooting"
    ]
}


# ============================================================
# 3. MODEL FEATURES
# ============================================================

features = [
    "skill_match_score",
    "distance_score",
    "availability_score",
    "rating_score",
    "experience_score",
    "certification_score"
]

class_names = {
    0: "Poor",
    1: "Moderate",
    2: "Good",
    3: "Excellent"
}


# ============================================================
# 4. MULTI-REQUEST RECOMMENDATION FUNCTION
# ============================================================

def get_recommendations(request, top_n=5):

    category = str(
        request["service_category"]
    ).strip()


    # --------------------------------------------------------
    # Filter workers
    # --------------------------------------------------------

    candidates = workers[
        workers["service_category"]
        .astype(str)
        .str.strip()
        .str.lower()
        ==
        category.lower()
    ].copy()


    if candidates.empty:
        return pd.DataFrame()


    # ========================================================
    # DISTANCE
    # ========================================================

    R = 6371.0

    customer_lat = float(
        request["customer_latitude"]
    )

    customer_lon = float(
        request["customer_longitude"]
    )

    lat1 = np.radians(customer_lat)
    lon1 = np.radians(customer_lon)

    lat2 = np.radians(
        pd.to_numeric(
            candidates["latitude"],
            errors="coerce"
        )
    )

    lon2 = np.radians(
        pd.to_numeric(
            candidates["longitude"],
            errors="coerce"
        )
    )

    dlat = lat2 - lat1
    dlon = lon2 - lon1

    a = (
        np.sin(dlat / 2) ** 2
        +
        np.cos(lat1)
        * np.cos(lat2)
        * np.sin(dlon / 2) ** 2
    )

    a = np.clip(a, 0, 1)

    c = 2 * np.arctan2(
        np.sqrt(a),
        np.sqrt(1 - a)
    )

    candidates["distance_km"] = R * c


    # ========================================================
    # DISTANCE SCORE
    # ========================================================

    max_distance = candidates[
        "distance_km"
    ].max()

    if pd.isna(max_distance) or max_distance == 0:

        candidates["distance_score"] = 1.0

    else:

        candidates["distance_score"] = (
            1
            -
            candidates["distance_km"]
            / max_distance
        ).clip(0, 1)


    # ========================================================
    # AVAILABILITY
    # ========================================================

    availability = (
        candidates["availability"]
        .astype(str)
        .str.strip()
        .str.lower()
    )

    candidates["availability_score"] = (
        availability.map({
            "available": 1.0,
            "unavailable": 0.0,
            "true": 1.0,
            "false": 0.0,
            "yes": 1.0,
            "no": 0.0
        }).fillna(0)
    )


    # ========================================================
    # RATING
    # ========================================================

    candidates["rating_score"] = (
        pd.to_numeric(
            candidates["rating"],
            errors="coerce"
        ) / 5.0
    ).clip(0, 1).fillna(0)


    # ========================================================
    # EXPERIENCE
    # ========================================================

    candidates["experience_years"] = (
        pd.to_numeric(
            candidates["experience_years"],
            errors="coerce"
        ).fillna(0)
    )

    max_exp = candidates[
        "experience_years"
    ].max()

    if max_exp > 0:

        candidates["experience_score"] = (
            candidates["experience_years"]
            / max_exp
        ).clip(0, 1)

    else:

        candidates["experience_score"] = 0.0


    # ========================================================
    # CERTIFICATION
    # ========================================================

    certification = (
        candidates["certification"]
        .astype(str)
        .str.strip()
        .str.lower()
    )

    candidates["certification_score"] = (
        certification.map({
            "certified": 1.0,
            "not certified": 0.0,
            "yes": 1.0,
            "no": 0.0,
            "true": 1.0,
            "false": 0.0
        }).fillna(0)
    )


    # ========================================================
    # SKILL MATCH
    # ========================================================

    required_skills = str(
        request["required_skills"]
    ).lower()

    required_skills = [
        x.strip()
        for x in required_skills.split(",")
        if x.strip()
    ]


    def calculate_skill_match(category):

        profile = service_skills.get(
            str(category).strip(),
            []
        )

        profile = [
            x.lower()
            for x in profile
        ]

        if len(required_skills) == 0:
            return 0.0

        matched = 0

        for required in required_skills:

            for skill in profile:

                if (
                    required in skill
                    or skill in required
                ):

                    matched += 1
                    break

        return matched / len(required_skills)


    candidates["skill_match_score"] = (
        candidates["service_category"]
        .apply(calculate_skill_match)
    )


    # ========================================================
    # PREPARE MODEL INPUT
    # ========================================================

    X_new = candidates[
        features
    ].copy()

    X_new = X_new.replace(
        [np.inf, -np.inf],
        np.nan
    ).fillna(0)


    # ========================================================
    # RANDOM FOREST PREDICTION
    # ========================================================

    candidates["predicted_class"] = (
        model.predict(X_new)
    )

    candidates["match_class"] = (
        candidates["predicted_class"]
        .map(class_names)
    )


    # ========================================================
    # CONFIDENCE
    # ========================================================

    if hasattr(model, "predict_proba"):

        probabilities = model.predict_proba(
            X_new
        )

        candidates["confidence"] = (
            np.max(
                probabilities,
                axis=1
            ) * 100
        )

    else:

        candidates["confidence"] = 0


    # ========================================================
    # FINAL MATCH SCORE
    # ========================================================

    candidates["final_match_score"] = (

        0.30 * candidates["skill_match_score"]

        + 0.20 * candidates["distance_score"]

        + 0.20 * candidates["availability_score"]

        + 0.15 * candidates["rating_score"]

        + 0.10 * candidates["experience_score"]

        + 0.05 * candidates["certification_score"]
    )


    candidates["match_percent"] = (
        candidates["final_match_score"] * 100
    )


    # ========================================================
    # SORT
    # ========================================================

    candidates = candidates.sort_values(
        by="final_match_score",
        ascending=False
    )


    # ========================================================
    # TOP N
    # ========================================================

    return candidates.head(top_n)


# ============================================================
# 5. TEST MULTIPLE REQUESTS
# ============================================================

NUMBER_OF_REQUESTS = 20

test_requests = requests.head(
    NUMBER_OF_REQUESTS
)


all_results = []


print("\n==========================================")
print("TESTING AI ON MULTIPLE REQUESTS")
print("==========================================")

print(
    "Number of customer requests:",
    len(test_requests)
)

print(
    "Workers evaluated per request:"
    " Top 5 selected"
)


# ============================================================
# 6. RUN ALL REQUESTS
# ============================================================

for _, request in test_requests.iterrows():

    result = get_recommendations(
        request,
        top_n=5
    )

    if not result.empty:

        result = result.copy()

        result["request_id"] = (
            request["request_id"]
        )

        result["request_service"] = (
            request["service_category"]
        )

        result["urgency"] = (
            request["urgency"]
        )

        all_results.append(result)


# ============================================================
# 7. COMBINE RESULTS
# ============================================================

if len(all_results) == 0:

    raise ValueError(
        "No recommendations were generated."
    )

multi_results = pd.concat(
    all_results,
    ignore_index=True
)


# ============================================================
# 8. SELECT IMPORTANT COLUMNS
# ============================================================

result_columns = [
    "request_id",
    "request_service",
    "urgency",
    "worker_id",
    "match_percent",
    "match_class",
    "confidence",
    "distance_km",
    "rating",
    "experience_years",
    "availability",
    "certification",
    "verification_status"
]

multi_results = multi_results[
    result_columns
].copy()


# Round values

multi_results["match_percent"] = (
    multi_results["match_percent"]
    .round(2)
)

multi_results["confidence"] = (
    multi_results["confidence"]
    .round(2)
)

multi_results["distance_km"] = (
    multi_results["distance_km"]
    .round(2)
)


# ============================================================
# 9. DISPLAY RESULTS
# ============================================================

print("\n==========================================")
print("MULTI-REQUEST AI RESULTS")
print("==========================================")

display(
    multi_results.head(25)
)


# ============================================================
# 10. PERFORMANCE SUMMARY
# ============================================================

print("\n==========================================")
print("AI RECOMMENDATION SUMMARY")
print("==========================================")

print(
    "Total requests tested:",
    test_requests.shape[0]
)

print(
    "Total recommendations generated:",
    len(multi_results)
)

print(
    f"Average match score: "
    f"{multi_results['match_percent'].mean():.2f}%"
)

print(
    f"Average confidence: "
    f"{multi_results['confidence'].mean():.2f}%"
)

print(
    f"Average distance: "
    f"{multi_results['distance_km'].mean():.2f} km"
)


# ============================================================
# 11. MATCH CLASS DISTRIBUTION
# ============================================================

print("\nMatch Class Distribution:")

class_distribution = (
    multi_results["match_class"]
    .value_counts()
)

display(
    class_distribution
    .rename_axis("Match Class")
    .reset_index(name="Count")
)


# ============================================================
# 12. SERVICE-WISE PERFORMANCE
# ============================================================

service_summary = (
    multi_results
    .groupby("request_service")
    .agg(
        recommendations=("worker_id", "count"),
        average_match=("match_percent", "mean"),
        average_distance=("distance_km", "mean"),
        average_rating=("rating", "mean")
    )
    .reset_index()
)

service_summary["average_match"] = (
    service_summary["average_match"]
    .round(2)
)

service_summary["average_distance"] = (
    service_summary["average_distance"]
    .round(2)
)

service_summary["average_rating"] = (
    service_summary["average_rating"]
    .round(2)
)

print("\n==========================================")
print("SERVICE-WISE AI PERFORMANCE")
print("==========================================")

display(
    service_summary
)


# ============================================================
# 13. MOST RECOMMENDED WORKERS
# ============================================================

worker_frequency = (
    multi_results["worker_id"]
    .value_counts()
    .head(10)
    .reset_index()
)

worker_frequency.columns = [
    "worker_id",
    "times_recommended"
]

print("\n==========================================")
print("MOST FREQUENTLY RECOMMENDED WORKERS")
print("==========================================")

display(
    worker_frequency
)


# ============================================================
# 14. GRAPH 1 - MATCH CLASS DISTRIBUTION
# ============================================================

plt.figure(figsize=(8, 5))

class_distribution.plot(
    kind="bar"
)

plt.title(
    "AI Recommendation Match Class Distribution"
)

plt.xlabel(
    "Match Class"
)

plt.ylabel(
    "Number of Recommendations"
)

plt.xticks(
    rotation=0
)

plt.tight_layout()

plt.show()


# ============================================================
# 15. GRAPH 2 - AVERAGE MATCH BY SERVICE
# ============================================================

plt.figure(figsize=(10, 5))

plt.bar(
    service_summary["request_service"],
    service_summary["average_match"]
)

plt.title(
    "Average AI Match Score by Service"
)

plt.xlabel(
    "Service Category"
)

plt.ylabel(
    "Average Match Score (%)"
)

plt.xticks(
    rotation=45,
    ha="right"
)

plt.tight_layout()

plt.show()


# ============================================================
# 16. GRAPH 3 - TOP RECOMMENDED WORKERS
# ============================================================

plt.figure(figsize=(9, 5))

plt.bar(
    worker_frequency["worker_id"].astype(str),
    worker_frequency["times_recommended"]
)

plt.title(
    "Most Frequently Recommended Workers"
)

plt.xlabel(
    "Worker ID"
)

plt.ylabel(
    "Number of Recommendations"
)

plt.xticks(
    rotation=45,
    ha="right"
)

plt.tight_layout()

plt.show()


# ============================================================
# 17. SAVE COMPLETE RESULTS
# ============================================================

multi_results.to_csv(
    "/content/multi_request_ai_results.csv",
    index=False
)

service_summary.to_csv(
    "/content/service_wise_ai_summary.csv",
    index=False
)

worker_frequency.to_csv(
    "/content/top_recommended_workers.csv",
    index=False
)


# ============================================================
# 18. FINAL STATUS
# ============================================================

print("\n==========================================")
print("STEP 9 COMPLETED SUCCESSFULLY")
print("==========================================")

print(
    "\nFiles created:"
)

print(
    "1. /content/multi_request_ai_results.csv"
)

print(
    "2. /content/service_wise_ai_summary.csv"
)

print(
    "3. /content/top_recommended_workers.csv"
)
```
